```python
import sqlite3

# 1. Connect to the database
db_path = "pubmed-vectors/pubmed_data.db"  # <--- REPLACE THIS with your file path
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# ---------------------------------------------------------
# Step A: Find your table name (If you don't know it)
# ---------------------------------------------------------
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()
print(f"Tables found in DB: {tables}")

# Let's assume the table is the first one found. 
# If you know the name, just set table_name = "YourTableName"
if tables:
    table_name = tables[0][0]
else:
    raise ValueError("No tables found in the database!")
```

```python
# 1. Count Total Rows
query_total = f"SELECT COUNT(*) FROM {table_name}"
cursor.execute(query_total)
total_rows = cursor.fetchone()[0]

# 2. Count Rows with Non-Empty Abstract
# We check for both NULL (missing) and empty strings ('')
query_abstract = f"""
    SELECT COUNT(*) 
    FROM {table_name} 
    WHERE Abstract IS NOT NULL AND Abstract != ''
"""
cursor.execute(query_abstract)
abstract_rows = cursor.fetchone()[0]

print(f"--- Results for table: {table_name} ---")
print(f"Total Rows: {total_rows}")
print(f"Rows with Abstract: {abstract_rows}")
print(f"Missing Abstracts: {total_rows - abstract_rows}")
```
